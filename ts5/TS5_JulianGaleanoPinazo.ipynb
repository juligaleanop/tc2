#### 1. Construir la Matriz Admitancia Indefinida para cada caso. En el BS170 plantear el modelo lineal visto en clase con admitancias y generadores de corriente, que tenga en cuenta $C_{iss}$, $C_{oss}$, $C_{rss}$ y $gm$.

##### Circuito T:

<img src="fig/circ_t.png"  width="300"/>

El procedimiento para obtener la Matriz Admitancia Indefinida es plantear las ecuaciones de cada nodo del circuito:

$$
\begin{cases}
I_1 = V_1(\frac{1}{sL}) - V_4(\frac{1}{sL}) \\
I_2 = V_2(\frac{1}{sL}) - V_4(\frac{1}{sL}) \\
I_3 = V_3(sC) - V_4(sC) \\
I_4 = V_4 (\frac{2}{sL} + sC) - V_1(\frac{1}{sL}) - V_2 (\frac{1}{sL}) - V_3 (sC)
\end{cases}
$$

Entonces la matriz queda:

$$
Y = \begin{bmatrix}
\frac{1}{sL} & 0 & 0 & -\frac{1}{sL} \\
0 & \frac{1}{sL} & 0 & -\frac{1}{sL} \\
0 & 0 & sC & -sC \\
-\frac{1}{sL} & -\frac{1}{sL} & -sC & \frac{s^2LC+2}{sL} 
\end{bmatrix}
$$

Se puede ver que se cumple la propiedad principal de las MAI, todas las filas y las columnas suman cero. Esta propiedad tiene como consecuencia las otras dos propiedades, los cofactores de primer orden son iguales y que la determinante de la matriz es igual a cero.

##### Circuito MOS equivalente:

<img src="fig/circ_mos.png"  width="500"/>

En este caso, plantear las ecuaciones de nodos requiere un poco mas de atención e incluir la corriente del generador controlado en los nodos que correspondan. Se abreviaran las capacitancias a $C_i$, $C_o$ y $C_r$ para simplificar, ya que se sobreentiende que son los valores dados por el fabricante para la configuración source común.

$$
\begin{cases}
I_1 = V_1(sC_i + sC_r) - V_2(sC_r) - V_3(sC_i) \\
I_2 = V_2(sC_o + sC_r) - V_1(sC_r) - V_3(sC_o) + gm(V_1+V_3) \\
I_3 = V_3(sC_i + sC_o) - V_1(sC_i) - V_2(sC_o) - gm(V_1+V_3) \\
\end{cases}
$$

Esto termina dando como resultado:

$$
Y = \begin{bmatrix}
s(C_i+C_r) & -sC_r & -sC_i \\
-sC_r+gm & s(C_o+C_r) & -sC_o-gm \\
-sC_i-gm & -sC_o & s(C_i+C_o)+gm \\
\end{bmatrix}
$$
Donde la fila/columna 1 corresponde al terminal de gate, la 2 corresponde al de drain y la 3 al de source.

#### 2. Para la red T se pide:

##### a. Obtener $\frac{V_{23}}{V_{13}}$, $\frac{V_{21}}{V_{31}}$ y $\frac{V_{12}}{V_{32}}$, explicitando los cofactores de segundo orden.

Las funciones de transferencia de tensión $T^{ij}_{mn}$ de una red multipolo puede calcularse a partir de la MAI de la siguiente forma:

$$ T^{ij}_{mn} = \frac{V_{ij}}{V_{mn}} = sgn(i-j)sgn(m-n)\frac{Y^{mn}_{ij}}{Y^{mn}_{mn}}$$

donde $Y^{mn}_{ij}$ es el cofactor de segundo orden que se obtiene del deteterminante eliminando las filas $m$ y $n$ y las columnas $i$ e $j$ y multiplicando por $(-1)^{i+j+m+n}$.

entonces, para calcular cada una de las transferencias resulta:

$$
T^{23}_{13} = sgn(2-3)sgn(1-3)\frac{Y^{13}_{23}}{Y^{13}_{13}},
$$
$$
Y^{13}_{23} = (-1)^{9}\begin{vmatrix} 0 & -\frac{1}{sL} \\ -\frac{1}{sL} & \frac{s^2LC+2}{sL} \end{vmatrix} = (-1)\cdot (-1) \frac{1}{s^2L^2} = \frac{1}{s^2L^2}
$$
$$
Y^{13}_{13} = (-1)^{8}\begin{vmatrix} \frac{1}{sL} & -\frac{1}{sL} \\ -\frac{1}{sL} & \frac{s^2LC+2}{sL} \end{vmatrix} = \frac{s^2LC+1}{s^2L^2} 
$$

$$
\Rightarrow T^{23}_{13} = (-1)(-1)\frac{\frac{1}{s^2L^2}}{\frac{s^2LC+1}{s^2L^2}} = \dfrac{\frac{1}{LC}}{s^2+\frac{1}{LC}}
$$

$$
T^{21}_{31} = sgn(2-1)sgn(3-1)\frac{Y^{31}_{21}}{Y^{31}_{31}},
$$
$$
Y^{31}_{21} = (-1)^{7}\begin{vmatrix} 0 & -\frac{1}{sL} \\ -sC & \frac{s^2LC+2}{sL} \end{vmatrix} = (-1)\cdot (-1) \frac{C}{L} = \frac{C}{L}
$$
$$
Y^{31}_{31} = (-1)^{8}\begin{vmatrix} \frac{1}{sL} & -\frac{1}{sL} \\ -\frac{1}{sL} & \frac{s^2LC+2}{sL} \end{vmatrix} = \frac{s^2LC+1}{s^2L^2} 
$$

$$
\Rightarrow T^{21}_{31} = \frac{\frac{C}{L}}{\frac{s^2LC+1}{s^2L^2}} = \dfrac{s^2}{s^2+\frac{1}{LC}}
$$

$$
T^{12}_{32} = sgn(1-2)sgn(3-2)\frac{Y^{32}_{12}}{Y^{32}_{32}},
$$
$$
Y^{32}_{12} = (-1)^{8}\begin{vmatrix} 0 & -\frac{1}{sL} \\ -sC & \frac{s^2LC+2}{sL} \end{vmatrix} = -\frac{C}{L}
$$
$$
Y^{32}_{32} = (-1)^{10}\begin{vmatrix} \frac{1}{sL} & -\frac{1}{sL} \\ -\frac{1}{sL} & \frac{s^2LC+2}{sL} \end{vmatrix} = \frac{s^2LC+1}{s^2L^2} 
$$

$$
\Rightarrow T^{12}_{32} = (-1)\frac{-\frac{C}{L}}{\frac{s^2LC+1}{s^2L^2}} = \dfrac{s^2}{s^2+\frac{1}{LC}}
$$

Lo cual tiene sentido, ya que la red es simétrica respecto al terminal 3 y en ambas transferencias se esta entrando por este.

Haciendo uso del cálculo simbólico provisto por el módulo _sympy_ y las funciones del módulo _pytc2_ para cuadripolos, verificaremos las transferencias obtenidas:

```python
import numpy as np
import sympy as syp
from pytc2.cuadripolos import calc_MAI_vtransf_ij_mn
from pytc2.general import print_latex

s, L, C = syp.symbols('s L C')
Ymai = syp.Matrix([  
                    [ 1/(s*L),  0,        0,    -1/(s*L)],
                    [ 0,        1/(s*L),  0,    -1/(s*L)],
                    [ 0,        0,        s*C,  -s*C    ],
                    [ -1/(s*L), -1/(s*L), -s*C, (s**2*L*C+2)/(s*L) ]
                 ])

#Se le resta 1 al indice de cada fila y columna ya que en las matrices de Sympy estos comienzan en 0.
Vmai1 = calc_MAI_vtransf_ij_mn(Ymai, 1, 2, 0, 2, verbose=False)
print_latex( r'T^{{ {:d}{:d} }}_{{ {:d}{:d} }} = '.format(2, 3, 1, 3) +  syp.latex(Vmai1) )

Vmai2 = calc_MAI_vtransf_ij_mn(Ymai, 1, 0, 2, 0, verbose=False)
print_latex( r'T^{{ {:d}{:d} }}_{{ {:d}{:d} }} = '.format(2, 1, 3, 1) +  syp.latex(Vmai2) )

Vmai3 = calc_MAI_vtransf_ij_mn(Ymai, 0, 1, 2, 1, verbose=False)
print_latex( r'T^{{ {:d}{:d} }}_{{ {:d}{:d} }} = '.format(1, 2, 3, 2) +  syp.latex(Vmai3) )
```

Se puede ver como los resultados obtenidos mediante cálculo simbólico coinciden con los obtenidos analíticamente.

#### 3. Para el MOS se pide:

##### a. Dada la MAI hallada en 1. obtener los parametros Y para las configuraciones source común y drain común.

Cada configuracion del transistor tiene una Matriz Impedancia Definida asociada, donde el respectivo terminal común es establecido como nodo de referencia. La forma de obtener una MAD a partir de la MAI de la red es simplemente eliminar la fila y columna asociada a ese nodo.

Por lo tanto, para la configuración source común la MAD es:

$$
Y_{ss} = \begin{bmatrix}
s(C_i+C_r) & -sC_r \\
-sC_r+gm & s(C_o+C_r)
\end{bmatrix}
$$

y la del drain común es:
$$
Y_{dd} = \begin{bmatrix}
s(C_i+C_r) & -sC_i \\
-sC_i-gm & s(C_i+C_o)+gm
\end{bmatrix}
$$

##### b. Calcular la matriz de parámetros ABCD en configuración source común.

La transformación de parametros desde Y a ABCD es la siguiente:

$$
A=\left. \frac{V_1}{V_2} \right|_{I_2=0} = -\frac{Y_{22}}{Y_{21}}
$$

$$
B=\left. \frac{V_1}{I_2} \right|_{V_2=0} = -\frac{1}{Y_{21}}
$$

$$
C=\left. \frac{I_1}{V_2} \right|_{I_2=0} = \frac{Y_{12}Y_{21}-Y{22}Y_{11}}{Y_{21}} = -\frac{\Delta Y}{Y_{21}}
$$

$$
D=\left. \frac{I_1}{I_2} \right|_{V_2=0} = -\frac{Y_{11}}{Y_{21}}
$$

$$
T = -\frac{1}{Y_{21}} \begin{bmatrix}
Y_{22} & 1 \\
\Delta Y & Y_{11}
\end{bmatrix} \Rightarrow
T_{ss} = \frac{1}{sC_r-gm} \begin{bmatrix}
s(C_o+C_r) & 1 \\
s^2(C_iC_o+C_iC_r+C_oC_r+\frac{gmC_r}{s}) & s(C_i+C_r)
\end{bmatrix}
$$
